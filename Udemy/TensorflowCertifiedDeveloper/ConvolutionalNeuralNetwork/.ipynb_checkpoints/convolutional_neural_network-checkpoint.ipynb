Importing the libraries

```python
import tensorflow as tf
from tensorflow import keras
import keras.preprocessing.image
from keras.utils import image_dataset_from_directory
print(tf.__version__)
```

Part 1 - Data Preprocessing

Preprocessing the Training set

```python
training_set = keras.utils.image_dataset_from_directory(
    directory="/home/singh/temp/udemyds/dataset/dataset/training_set",
    labels="inferred",       # Automatically assign labels from folder names
    label_mode="int",        # Can also be "categorical" or "binary"
    batch_size=32,
    image_size=(200, 200),
    shuffle=True
)
```

Preprocessing the Test set

```python
test_set = keras.utils.image_dataset_from_directory(
    directory="/home/singh/temp/udemyds/dataset/dataset/test_set",
    labels="inferred",       # Automatically assign labels from folder names
    label_mode="int",        # Can also be "categorical" or "binary"
    batch_size=32,
    image_size=(200, 200),
    shuffle=True
) 
```

Part 2 - Building the CNN

Initialising the CNN

```python
cnn = tf.keras.models.Sequential()
```

Step 1 - Convolution

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation="relu", input_shape=[64, 64, 3]))
```

Step 2 - Pooling

```python
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```
