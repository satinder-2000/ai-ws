Multinomial distribution is a generalization of binomial distribution.

It describes outcomes of multi-nomial scenarios unlike binomial where scenarios must be only one of two. e.g. Blood type of a population, dice roll outcome.

It has three parameters:
n - number of times to run the experiment.
pvals - list of probabilties of outcomes (e.g. [1/6, 1/6, 1/6, 1/6, 1/6, 1/6] for dice roll).
size - The shape of the returned array.

IB:
In investment banking, you can use the multinomial distribution to estimate the probability of a specific mix of deal or market outcomes across multiple independent transactions—such as 3 Strong closes (Full advisory fee plus performance kicker), 2 Steady closes(Base advisory fee at target timeline ), and 1 Weak close(Discounted fee or delayed/re-cut terms) out of 6 advisory mandates, which has a 13.50% probability.

```python
from numpy import random

x = random.multinomial(n=6, pvals=[1/6, 1/6, 1/6, 1/6, 1/6, 1/6])
x
```
