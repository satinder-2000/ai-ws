Chi Square distribution is used as a basis to verify the hypothesis.

It has two parameters:
df - (degree of freedom).
size - The shape of the returned array.

IB
In investment banking, a chi-square test evaluates whether categorical deal outcomes differ significantly from expectations—for example, testing whether M&A deal completion is independent of financing structure (cash vs. stock/debt) across 120 analyzed transactions.

```python
from numpy import random
x = random.chisquare(df=2, size=(2,3))
x
```

<h3>Visualization of Chi Square Distribution</h3>

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.chisquare(df=1, size=1000), kind="kde")
plt.show()
```
