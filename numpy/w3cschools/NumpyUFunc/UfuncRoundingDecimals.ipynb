There are primarily five ways of rounding off decimals in NumPy:

truncation
fix
rounding
floor
ceil

<h3>Truncation</h3>

```python
import numpy as np
```

```python
arr = np.trunc([-3.1667, 3.5667])
arr
```

```python
arr = np.fix([-3.1667, 3.5667])
arr
```

<h3>Rounding</h3>
around

```python
arr = np.around(3.1666, 2)
arr
```

<h3>Floor</h3>

```python
arr = np.floor([3.1666, 2.6666])
arr
```

<h3>Ceil</h3>

```python
arr = np.ceil([3.1666, 2.6666])
arr
```
