## Ufunc methods
All ufuncs have 5 methods. 4 reduce-like methods (reduce, accumulate, reduceat, outer) and one for inplace operations (at). 

```python
np.add.reduce([1,2,3])
```

```python
np.divmod.reduce([1,2,3])
```

```python
x = np.arange(9).reshape(3, 3)
x
```

```python
np.add.reduce(x, 1)
```

```python
np.add.reduce(x, (0, 1))
```

```python
x.dtype
```

```python
np.multiply.reduce(x, dtype=np.float64)
```
